```python
#1. Why does this feature selection with chi2 fail?
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.preprocessing import MinMaxScaler

X = [[-1, 2, 3], [4, 0, -6], [7, 8, 9], [1, 2, 3]]
y = [0, 1, 0, 1]

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)


selector = SelectKBest(score_func=chi2, k=2)
X_new = selector.fit_transform(X_scaled, y)

print(X_new)
```

```python
#2.
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import CountVectorizer

# Error: strings instead of numerical features
user1 = 'action comedy romance'
user2 = 'action thriller comedy'

vetorizer = CountVectorizer()

X = vetorizer.fit_transform([user1, user2])

similarity = cosine_similarity(X[0], X[1])

print(similarity)
```

```python
#5.
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression
import warnings
warnings.filterwarnings("ignore")

X = ([1,3,4], [4,5,0], [9,7,48])
y = (2,3,5)

selector = RFE(estimator=LogisticRegression(), n_features_to_select=2)
selector.fit(X, y)  # Error: X and y not defined

print(selector.support_)
print(selector.ranking_)
```

```python
#9.
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest, chi2


X = [[1, 2], [3, None], [4, 5]]
y = [0, 1, 0]

pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy = 'mean')),
    ('select', SelectKBest(score_func=chi2, k=1))
])

pipeline.fit(X, y)

print(pipeline.named_steps['select'].get_support())
```

```python
#10.
from scipy.sparse import csr_matrix
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.preprocessing import MinMaxScaler

X = csr_matrix([[0, -1, 3], [0, 2, 1], [3, -1, 0]])
y = [0, 1, 0]

scaler = MinMaxScaler()
x_scaled = scaler.fit_transform(X.toarray())

selector = SelectKBest(score_func=chi2, k=2)
X_new = selector.fit_transform(x_scaled, y)

print(X_new)
```
