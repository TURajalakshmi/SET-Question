```python
#1. Use Mutual Information to find the most important feature for predicting Purchased. Select the top 1 feature.
import pandas as pd
from sklearn.feature_selection import SelectKBest, mutual_info_classif
import warnings
warnings.filterwarnings("ignore")

df = pd.DataFrame({
    'Age': [25, 32, 47, 51, 62],
    'Salary': [40000, 50000, 60000, 52000, 62000],
    'Purchased': [0, 1, 1, 0, 1]
})

X = df[['Age', 'Salary']]
y = df['Purchased']

def mi_score(X, y):
    return mutual_info_classif(X, y, random_state=42, n_jobs=1)

selector = SelectKBest(score_func=mi_score, k=1)

X_new = selector.fit_transform(X, y)

selected_feature = X.columns[selector.get_support()].tolist()

print(selected_feature)
```

```python
#2. Based on item popularity (sum of ratings), recommend top 2 items to a new user.
import pandas as pd

data = pd.DataFrame({
    'User': ['A', 'A', 'B', 'B', 'C', 'C'],
    'Item': ['Apple', 'Banana', 'Apple', 'Orange', 'Banana', 'Orange'],
    'Rating': [5, 3, 4, 2, 5, 4]
})

popularity = data.groupby('Item')['Rating'].sum()

top_2 = popularity.sort_values(ascending=False).head(2).index.tolist()

print(top_2)
```

```python
#3. Write a Python program to perform feature selection using Recursive Feature Elimination (RFE). 
# Use Linear Regression as the estimator and select the top 5 features. Print the selected feature names.

from sklearn.datasets import load_diabetes
import pandas as pd
from sklearn.feature_selection import RFE
from sklearn.linear_model import LinearRegression

X, y = load_diabetes(return_X_y=True, as_frame=True)

model = LinearRegression()

rfe = RFE(estimator=model, n_features_to_select=5)
rfe.fit(X,y)

selected_features = X.columns[rfe.support_].tolist()

print("Selected features:", selected_features)
```

```python
#4. Write a Python program that selects the top 8 features using SelectKBest with mutual_info_classif as the scoring function. 
# Print the names of the selected features.

from sklearn.datasets import load_digits
import pandas as pd
from sklearn.feature_selection import SelectKBest, mutual_info_classif

X, y = load_digits(return_X_y=True, as_frame=True)

Selector = SelectKBest(score_func=mutual_info_classif, k=8)
Selector.fit(X,y)

selected_feature = X.columns[Selector.get_support()].tolist()
print("Top 8 features using mutual info:", selected_feature)
```

```python
#5. Write a Python program to perform feature selection using SelectFromModel. 
# Train a Logistic Regression model and select features based on importance using SelectFromModel. Print the names of the selected feature

import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import SelectFromModel

X, y = load_breast_cancer(return_X_y=True, as_frame=True)

model = LogisticRegression(max_iter=10000)
model.fit(X, y)

selector = SelectFromModel(model, prefit=True)

Selected_features = X.columns[selector.get_support()].tolist()

print("Selected features:", Selected_features)
```
